```python
# Cell 2: Load Document Corpus
print("="*80)
print("📂 LOADING DOCUMENT CORPUS")
print("="*80)

# Load document corpus
df = pd.read_csv('../data/document_corpus_enhanced.csv')
print(f"✅ Loaded {len(df):,} documents")

# Generate multimodal embeddings (simulating LayoutLMv3 output)
np.random.seed(42)
embedding_dim = 384  # Simulating LayoutLMv3 hidden dimension

# Text embeddings (simulated)
text_embeddings = np.random.randn(len(df), embedding_dim)

# Layout embeddings (position-aware)
layout_embeddings = np.random.randn(len(df), embedding_dim) * 0.8

# Image embeddings (visual features)
image_embeddings = np.random.randn(len(df), embedding_dim) * 0.6

# Fused embeddings (weighted combination)
fused_embeddings = 0.5 * text_embeddings + 0.3 * layout_embeddings + 0.2 * image_embeddings

print(f"✅ Generated multimodal embeddings:")
print(f"   Text embeddings: {text_embeddings.shape}")
print(f"   Layout embeddings: {layout_embeddings.shape}")
print(f"   Image embeddings: {image_embeddings.shape}")
print(f"   Fused embeddings: {fused_embeddings.shape}")
```

```python
# Cell 2: Generate Short Drama Dataset 
print("="*70)
print("📊 GENERATING SHORT DRAMA DATASET")
print("="*70)

from datetime import datetime, timedelta

n_dramas = 10000
start_date = datetime(2024, 1, 1)

# Define categories
genres = ['Romance', 'Action', 'Comedy', 'Drama', 'Thriller', 'Horror', 'Sci-Fi', 'Fantasy']
descriptions = [
    'A thrilling love story set in modern Seoul with unexpected twists',
    'Historical romance during the Joseon dynasty filled with intrigue',
    'Action-packed martial arts adventure with breathtaking fight scenes',
    'Heartwarming family comedy about relationships and laughter',
    'Mystery thriller with unexpected twists and shocking revelations',
    'Coming-of-age drama about friendship, love, and growing up',
    'Horror suspense with supernatural elements and jump scares',
    'Science fiction dystopian future exploring human nature'
]

# Create dataframe
drama_data = pd.DataFrame({
    'content_id': range(1, n_dramas + 1),
    'title': [f'Short Drama Episode {i}' for i in range(1, n_dramas + 1)],
    'description': np.random.choice(descriptions, n_dramas),
    'genre': np.random.choice(genres, n_dramas),
    'rating': np.random.uniform(1, 5, n_dramas).round(1),
    'view_count': np.random.poisson(50000, n_dramas),
    'like_count': np.random.poisson(5000, n_dramas),
    'comment_count': np.random.poisson(500, n_dramas),
    'share_count': np.random.poisson(200, n_dramas),
    'upload_date': [start_date + timedelta(hours=i) for i in range(n_dramas)]
})

# Save to CSV
drama_data.to_csv('drama_content.csv', index=False)

print(f"✅ Generated {len(drama_data)} short drama episodes")
print(f"   Saved to: drama_content.csv")
print(f"\n📊 Sample data:")
print(drama_data.head(3))
print(f"\n📊 Drama columns: {drama_data.columns.tolist()}")
```

```python
# Cell 3: Generate Photo-Text Dataset
print("="*70)
print("📷 GENERATING PHOTO-TEXT DATASET")
print("="*70)

import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

np.random.seed(42)
random.seed(42)

n_photos = 10000
start_date = datetime(2024, 1, 1)

# Define categories
image_categories = ['Nature', 'City', 'Food', 'People', 'Animals', 'Art', 'Technology', 'Fashion']
captions = [
    'a beautiful sunset over the ocean with golden reflections',
    'a busy city street at night with bright neon lights',
    'a peaceful mountain landscape with snow-capped peaks',
    'a delicious plate of gourmet food beautifully presented',
    'a cute dog playing in the park on a sunny day',
    'a modern architectural building with unique design',
    'a colorful abstract painting with vibrant expressions',
    'a group of friends laughing together at a cafe'
]

# Create dataframe
photo_data = pd.DataFrame({
    'image_id': range(1, n_photos + 1),
    'caption': np.random.choice(captions, n_photos),
    'category': np.random.choice(image_categories, n_photos),
    'engagement_score': np.random.uniform(0, 100, n_photos).round(1),
    'like_count': np.random.poisson(1000, n_photos),
    'comment_count': np.random.poisson(100, n_photos),
    'share_count': np.random.poisson(50, n_photos),
    'upload_date': [start_date + timedelta(hours=i) for i in range(n_photos)],
    'uploader_id': [f'user_{np.random.randint(1, 1001)}' for _ in range(n_photos)]
})

# Save to CSV
photo_data.to_csv('photo_content.csv', index=False)

print(f"✅ Generated {len(photo_data)} photo-text items")
print(f"   Saved to: photo_content.csv")
print(f"\n📷 Sample data:")
print(photo_data.head(3))
```

```python
# Cell 4: Generate User Search Query Logs
print("="*70)
print("🔍 GENERATING USER SEARCH QUERY LOGS")
print("="*70)

import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

np.random.seed(42)
random.seed(42)

n_queries = 50000
start_date = datetime(2024, 1, 1)

# Define categories
genres = ['Romance', 'Action', 'Comedy', 'Drama', 'Thriller', 'Horror', 'Sci-Fi', 'Fantasy']
search_intents = ['Informational', 'Navigational', 'Transactional', 'Entertainment', 'Social']

search_queries = []

for i in range(n_queries):
    category = np.random.choice(genres + ['short drama', 'video', 'content'])
    query_text = np.random.choice([
        f'best {category} content',
        f'{category} short drama',
        f'watch {category} online',
        f'top rated {category} series'
    ])
    
    search_queries.append({
        'query_id': i + 1,
        'query_text': query_text,
        'search_intent': np.random.choice(search_intents),
        'user_id': f'user_{np.random.randint(1, 5001)}',
        'timestamp': start_date + timedelta(hours=np.random.randint(0, 10000)),
        'device_type': np.random.choice(['Mobile', 'Desktop', 'Tablet'], p=[0.6, 0.3, 0.1])
    })

query_logs = pd.DataFrame(search_queries)
query_logs.to_csv('query_logs.csv', index=False)

print(f"✅ Generated {len(query_logs)} search queries")
print(f"   Saved to: query_logs.csv")
print(f"\n🔍 Sample queries:")
print(query_logs.head(5))
```

```python
# Cell 5: Generate Relevance Judgments (FASTER VERSION)
print("="*70)
print("🏷️ GENERATING RELEVANCE JUDGMENTS (GROUND TRUTH)")
print("="*70)

import pandas as pd
import numpy as np
import os

np.random.seed(42)

# Load the previously saved data
drama_data = pd.read_csv('drama_content.csv')
query_logs = pd.read_csv('query_logs.csv')

print(f"✅ Loaded drama_data: {len(drama_data)} records")
print(f"✅ Loaded query_logs: {len(query_logs)} records")

n_judgments = 100000

# Vectorized approach - MUCH FASTER
query_ids = np.random.choice(query_logs['query_id'].values, n_judgments)
content_ids = np.random.choice(drama_data['content_id'].values, n_judgments)
relevance_scores = np.random.choice([0, 1, 2, 3, 4], n_judgments, p=[0.3, 0.25, 0.2, 0.15, 0.1])

judgments_df = pd.DataFrame({
    'query_id': query_ids,
    'content_id': content_ids,
    'relevance_score': relevance_scores,
    'judgment_source': 'simulated'
})

judgments_df.to_csv('relevance_judgments.csv', index=False)

print(f"✅ Generated {len(judgments_df)} relevance judgments")
print(f"   Saved to: relevance_judgments.csv")
print(f"\n📊 Relevance score distribution:")
print(judgments_df['relevance_score'].value_counts().sort_index())
```
